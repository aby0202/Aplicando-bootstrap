### ✅ Solución ejercicio 2

````python
sql_query_a = """
SELECT
    cliente_id,
    COUNT(pedido_id)    AS frecuencia,
    SUM(monto_total)    AS ltv,
    AVG(monto_total)    AS ticket_promedio,
    MIN(fecha_venta)    AS primera_compra
FROM ventas
GROUP BY cliente_id
HAVING COUNT(pedido_id) > 2;
"""

sql_query_b = """
SELECT
    cliente_id,
    COUNT(pedido_id)    AS frecuencia,
    SUM(monto_total)    AS ltv,
    AVG(monto_total)    AS ticket_promedio,
    MIN(fecha_venta)    AS primera_compra
FROM ventas
WHERE fecha_venta >= '2025-01-01'
GROUP BY cliente_id
HAVING COUNT(pedido_id) > 2;
"""

sql_query_c = """
SELECT
    cliente_id,
    COUNT(pedido_id)                    AS frecuencia,
    SUM(monto_total)                    AS ltv,
    AVG(monto_total)                    AS ticket_promedio,
    MIN(fecha_venta)                    AS primera_compra,
    CASE
        WHEN SUM(monto_total) > 5000  THEN 'Alto'
        WHEN SUM(monto_total) >= 1000 THEN 'Medio'
        ELSE 'Bajo'
    END                                 AS segmento
FROM ventas
WHERE fecha_venta >= '2025-01-01'
GROUP BY cliente_id
HAVING COUNT(pedido_id) > 2;
"""

# You can now print these queries or execute them using a SQL connector.
# For example:
# print("Query a:")
# print(sql_query_a)
# print("\nQuery b:")
# print(sql_query_b)
# print("\nQuery c:")
# print(sql_query_c)
````

### ✅ Solución ejercicio 3

````python
# a) Estandarizar categoria
df['categoria'] = df['categoria'].str.lower().str.strip()
print(df['categoria'].value_counts())
# Resultado: 3 categorías limpias: ropa, electrónica, hogar

# b) Imputar edad con mediana
mediana_edad = df['edad'].median()
df['edad'] = df['edad'].fillna(mediana_edad)
print(f"Mediana usada: {mediana_edad}")
print(df['edad'].isnull().sum())  # → 0

# c) Detectar outliers en salario con IQR
Q1 = df['salario'].quantile(0.25)
Q3 = df['salario'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['salario'] < Q1 - 1.5 * IQR) | (df['salario'] > Q3 + 1.5 * IQR)]
print("Outliers detectados:")
print(outliers[['cliente_id', 'salario']])
# cliente 8 con salario 200000 debería aparecer

# d) Parsear fechas con formato mixto
df['fecha_reg'] = pd.to_datetime(df['fecha_reg'], format='mixed', dayfirst=True)
print(df['fecha_reg'].dtype)  # → datetime64[ns]

# e) Verificar y exportar
print("Nulos restantes:")
print(df.isnull().sum())
df.to_csv('clientes_limpios.csv', index=False)
print("✅ Dataset exportado sin nulos")
````
